```python
eligible_pop_cont = eligible_pop[(eligible_pop['Year'] >= 1995) & (eligible_pop['Year'] <= 2022)].copy()
eligible_pop_cont.head()
```

```python
continent_players['season_start'] = (
    continent_players['season']
    .str[:4]
    .astype(int)
)

continent_players.head()
```

```python
# Prepare population dataset
population_merge = eligible_pop_cont.rename(columns={
    'Year': 'year',
    'Continent': 'continent',
    'Population': 'total_pop'
})

# Merge population with player counts
continent_analysis = population_merge.merge(
    continent_players,
    left_on=['year', 'continent'],
    right_on=['season_start', 'continent'],
    how='left'
)

continent_analysis.head()
```
